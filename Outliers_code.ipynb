```python
import numpy as np
np.random.seed(42)
n=30
sensor_data = np.random.normal(loc=25, scale=2, size=n)
sensor_data = np.round(sensor_data)
print(sensor_data)
```

```python
sensor_data[6]=11
sensor_data[17]=78
sensor_data[22]=56
print("Sensor_Data:\n",sensor_data)
```

```python
mean = np.mean(sensor_data)
print("Mean:\n",mean)
```

```python
std = np.std(sensor_data)
print("Standard_Deviation:\n",std)
```
